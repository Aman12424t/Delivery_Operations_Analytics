# SwiftShip Delivery Operations — Data Cleaning

This notebook cleans and prepares the raw delivery operations dataset for analysis by handling duplicates, missing values, invalid values, inconsistent dates, and data types.

```python
import numpy as np 
import pandas as pd
```

## 1. Load the Raw Dataset

Load the raw delivery operations dataset that was identified and inspected during the data understanding stage.

```python
df = pd.read_csv("../Data/Raw/swiftship_delivery_operations_raw.csv")
```

```python
df.head()
```

## 2. Remove Duplicate Records

Remove duplicate records and verify that no duplicate rows remain in the dataset.

```python
df.drop_duplicates(inplace = True)
```

```python
df.duplicated().sum()
```

```python
df
```

## 3. Handle Missing Values

Identify missing values and apply appropriate replacements to categorical and descriptive fields where required.

```python
df.isnull().sum()
```

```python
df["customer_segment"] = df["customer_segment"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["delivery_partner"] = df["delivery_partner"].fillna("delivery_partner")
df["weather_condition"] = df["weather_condition"].fillna("Unknown")
df["return_reason"] = df["return_reason"].fillna("Not Provided")
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

## 4. Handle Invalid Numerical Values

Remove records containing invalid numerical values such as non-positive quantities, order values, distances, package weights, and invalid customer ratings.

```python
df = df[df["quantity"] > 0]
```

```python
(df["quantity"]<=0).sum()
```

```python
df.info()
```

```python
df["customer_rating"] = pd.to_numeric(
    df["customer_rating"],
    errors="coerce"
)
```

```python
df = df[df["order_value"]>0] 
df = df[df["distance_km"]>0]
df = df[df["package_weight_kg"]>0]
df = df[(df["customer_rating"] > 0) & (df["customer_rating"] <= 5)]
```

```python
df["delivery_partner"] = df["delivery_partner"].replace("delivery_partner", "Unknown")
```

## 5. Validate Delivery Dates

Identify and remove records where the delivery timeline is logically inconsistent, such as dispatch occurring before the order or delivery occurring before dispatch.

```python
invalid_rows =  ( (df["dispatch_date"] < df["order_date"]) |
                (df["delivery_date"] < df["dispatch_date"])|
                 (df["delivery_date"] < df["order_date"] ))

invalid_rows.sum()
```

```python
df = df[~invalid_rows]
```

## 6. Convert Date Columns

Convert order, dispatch, promised delivery, and actual delivery dates into Pandas datetime format for reliable date-based analysis.

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["delivery_date"] = pd.to_datetime(df["delivery_date"])
df["dispatch_date"] = pd.to_datetime(df["dispatch_date"])
df["promised_delivery_date"] = pd.to_datetime(df["promised_delivery_date"])
```
